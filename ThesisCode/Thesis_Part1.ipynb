# Rebuilding Atomic Orbitals & Reorganizing the Periodic Table

**Part 1.Rebuilding Atomic Orbitals**

```python
import numpy as np
import plotly.graph_objects as go
from plotly.io import to_html

SAFE_NS = {
    "np": np,
    "sin": np.sin, "cos": np.cos, "tan": np.tan,
    "sqrt": np.sqrt, "abs": np.abs,
    "exp": np.exp, "log": np.log,
    "pi": np.pi
}

def make_sphere_grid(n_theta=100, n_phi=100):
    theta = np.linspace(0, np.pi, n_theta)
    phi   = np.linspace(0, 2*np.pi, n_phi)
    theta, phi = np.meshgrid(theta, phi)
    x = np.sin(theta) * np.cos(phi)
    y = np.sin(theta) * np.sin(phi)
    z = np.cos(theta)
    return x, y, z


def make_orbital(title, expr_str, n_theta=80, n_phi=80,
                 colorscale="Viridis", showscale=False, scale=0.85):
    x, y, z = make_sphere_grid(n_theta, n_phi)
    local_ns = {"x": x, "y": y, "z": z}
    f_val = eval(expr_str, {"__builtins__": {}}, {**SAFE_NS, **local_ns})

    f_arr = np.asarray(f_val)
    if f_arr.shape == ():
        f_arr = np.ones_like(x) * float(f_arr)
    else:
        f_arr = f_arr.astype(float, copy=False)


    R_raw = np.abs(f_arr)
    r_min, r_max = R_raw.min(), R_raw.max()


    if np.isclose(r_max, r_min):
        R_norm = np.ones_like(R_raw) * 0.6
    else:
        R_norm = (R_raw - r_min) / (r_max - r_min + 1e-12)

    R = R_norm * scale
    X, Y, Z = R * x, R * y, R * z

    Rn = R_norm

    fixed_axis_len = 1.2
    axis_color, arrow_color, lw = "#444444", "#222222", 2.5
    fixed_arrow_size = 0.12

    fig = go.Figure(data=[go.Surface(
        x=X, y=Y, z=Z,
        surfacecolor=Rn, cmin=0, cmax=1,
        colorscale=colorscale, showscale=showscale,
        hoverinfo='none'
    )])

    for end_pt in [(fixed_axis_len, 0, 0), (0, fixed_axis_len, 0), (0, 0, fixed_axis_len)]:
        fig.add_trace(go.Scatter3d(
            x=[0, end_pt[0]], y=[0, end_pt[1]], z=[0, end_pt[2]],
            mode="lines", line=dict(color=axis_color, width=lw),
            showlegend=False, hoverinfo="skip"
        ))


    def add_arrow(tip, vec):
        fig.add_trace(go.Cone(
            x=[tip[0]], y=[tip[1]], z=[tip[2]],
            u=[vec[0]], v=[vec[1]], w=[vec[2]],
            anchor="tip",
            sizemode="absolute",
            sizeref=fixed_arrow_size,
            showscale=False,
            colorscale=[[0, arrow_color], [1, arrow_color]],
            hoverinfo="skip"
        ))

    add_arrow((fixed_axis_len, 0, 0), (1, 0, 0))
    add_arrow((0, fixed_axis_len, 0), (0, 1, 0))
    add_arrow((0, 0, fixed_axis_len), (0, 0, 1))


    label_off = fixed_axis_len + 0.15
    fig.add_trace(go.Scatter3d(
        x=[label_off, 0, 0], y=[0, label_off, 0], z=[0, 0, label_off],
        mode="text", text=["x", "y", "z"],
        textfont=dict(color=axis_color, size=10),
        showlegend=False, hoverinfo="skip"
    ))


    limit = fixed_axis_len * 1.25
    axis_style = dict(
        showbackground=False, showgrid=False, showline=False,
        showticklabels=False, ticks='', title='', visible=False,
        range=[-limit, limit]
    )

    fig.update_layout(
        scene=dict(
            xaxis=axis_style, yaxis=axis_style, zaxis=axis_style,
            aspectmode="cube",
            camera=dict(eye=dict(x=1.8, y=1.8, z=1.4))
        ),
        margin=dict(l=0, r=0, t=0, b=0),
        paper_bgcolor='rgba(0,0,0,0)',
        plot_bgcolor='rgba(0,0,0,0)'
    )
    return title, fig


rows = [
    ("s orbitals", [
        make_orbital(r"s", "1")
    ]),
    ("p orbitals", [
        make_orbital(r"p_x", "x"),
        make_orbital(r"p_y", "y"),
        make_orbital(r"p_z", "z"),
    ]),
    ("d orbitals", [
        make_orbital(r"d_{z^2}", "2*z**2 - x**2 - y**2"),
        make_orbital(r"d_{x^2-y^2}", "x**2 - y**2"),
        make_orbital(r"d_{xy}", "x*y"),
        make_orbital(r"d_{xz}", "x*z"),
        make_orbital(r"d_{yz}", "y*z"),
    ]),
    ("f orbitals", [
        make_orbital(r"f_{xyz}", "x*y*z"),
        make_orbital(r"f_{xz^2}", "3*x*(-4*z**2 + x**2 + y**2)"),
        make_orbital(r"f_{yz^2}", "3*y*(-4*z**2 + x**2 + y**2)"),
        make_orbital(r"f_{z(x^2-y^2)}", "z*(x**2 - y**2)"),
        make_orbital(r"f_{y(3x^2-y^2)}", "y*(3*x**2 - y**2)"),
        make_orbital(r"f_{z^3}", "z*(3*x**2+3*y**2-2*z**2)"),
        make_orbital(r"f_{x(x^2-3y^2)}", "x*(x**2 - 3*y**2)"),
    ]),
]


def fig_div(fig):
    return to_html(fig, include_plotlyjs=False, full_html=False,
                   default_width="100%", default_height="100%",
                   config={"responsive": True})

sections_html = []
for layer_name, items in rows:
    cards = []
    for title, fig in items:
        cards.append(f"""
        <div class="card">
          <div class="title">\\({title}\\)</div>
          <div class="plot">{fig_div(fig)}</div>
        </div>
        """)
    sections_html.append(f"""
    <section class="row">
      <div class="row-cards">
        {''.join(cards)}
      </div>
      <div class="row-label">{layer_name}</div>
    </section>
    """)

html = f"""<!doctype html>
<meta charset="utf-8">
<title>Orbital Pyramid</title>


<script src="https://cdn.plot.ly/plotly-latest.min.js"></script>
<script defer src="https://cdn.jsdelivr.net/npm/mathjax@3/es5/tex-chtml.js"></script>

<style>
  :root {{
    --card-w: 160px;
    --card-h: 190px;
    --maxw:   1400px;
  }}
  * {{ box-sizing: border-box; }}
  body {{
    margin: 0; padding: 24px 12px;
    font-family: system-ui,-apple-system,Segoe UI,Roboto,Helvetica,Arial;
    background: #f7f7f7;
  }}
  h1 {{
    margin: 6px 0 18px; text-align: center; font-weight: 700; letter-spacing: .3px;
  }}
  .pyramid {{
    max-width: var(--maxw);
    margin: 0 auto;
    display: flex;
    flex-direction: column;
    gap: 22px;
  }}


  .row {{
    display: flex;
    flex-direction: column;
    align-items: center;
    gap: 8px;
  }}
  .row-cards {{
    display: flex;
    flex-wrap: wrap;
    justify-content: center;
    gap: 14px;
    width: 100%;
  }}
  .row-label {{
    font-size: 14px; color: #444; text-align: center; opacity: .85;
    margin-top: 4px;
  }}


  .card {{
    width: min(18vw, var(--card-w));
    min-width: var(--card-w);
    background: #fff; border: 1px solid #eee; border-radius: 14px;
    padding: 8px 8px 10px;
    display: flex; flex-direction: column; align-items: stretch;
    overflow: hidden; clip-path: inset(0 round 14px);
    box-shadow: 0 1px 0 rgba(0,0,0,.02);
    transition: box-shadow .15s ease;
  }}
  .card:hover {{ box-shadow: 0 6px 18px rgba(0,0,0,.08); }}
  .title {{
    text-align: center; color:#243447; font-weight: 600;
    line-height: 1.2; min-height: 22px; margin-bottom: 6px;
  }}
  .plot {{
    height: var(--card-h);
    position: relative; overflow: hidden; isolation: isolate;
  }}
  .plot > div {{
    position: absolute !important; inset: 0 !important;
    width: 100% !important; height: 100% !important;
    overflow: hidden !important; transform: translateZ(0);
  }}
  .plot .gl-container, .plot .svg-container, .plot canvas {{
    position: absolute !important; inset: 0 !important;
    width: 100% !important; height: 100% !important;
    overflow: hidden !important; clip-path: inset(0) !important;
    transform: translateZ(0);
  }}
  .plot .hoverlayer {{ clip-path: inset(0) !important; }}
  .plot .modebar-container, .plot .modebar {{ overflow: hidden !important; }}


  @media (max-width: 900px) {{
    :root {{ --card-w: 140px; --card-h: 160px; }}
  }}
  @media (max-width: 600px) {{
    :root {{ --card-w: 120px; --card-h: 140px; }}
  }}
</style>

<h1>Interactive Atomic Orbital Pyramid</h1>
<div class="pyramid">
  {''.join(sections_html)}
</div>


<script>
(function() {{
  function resizeAll() {{
    document.querySelectorAll('.plot .js-plotly-plot').forEach(function(el) {{
      try {{ Plotly.Plots.resize(el); }} catch(e) {{}}
    }});
  }}
  window.addEventListener('load', resizeAll);
  const pyr = document.querySelector('.pyramid');
  if (pyr && 'ResizeObserver' in window) {{
    const ro = new ResizeObserver(resizeAll);
    ro.observe(pyr);
  }}
  if (window.MathJax && MathJax.startup) {{
    MathJax.startup.promise.then(resizeAll);
  }}
  setTimeout(resizeAll, 120);
  setTimeout(resizeAll, 400);
  setTimeout(resizeAll, 900);
}})();
</script>
"""

with open("orbitals_pyramid_new.html", "w", encoding="utf-8") as f:
    f.write(html)


```

**Part 2.Reorganizing Periodic Table**

```python
import numpy as np
import plotly.graph_objects as go


N_THETA = 32
N_PHI = 32


CONFIG_ANOMALIES = {
    'Cr': 5, 'Cu': 10,
    'Nb': 4, 'Mo': 5, 'Ru': 7, 'Rh': 8, 'Pd': 10, 'Ag': 10,
    'Pt': 9, 'Au': 10,
    'Gd': 7, 'Cm': 7,
    'La': 1, 'Ac': 1
}


ALL_ELEMENTS = [
    "", "H", "He", "Li", "Be", "B", "C", "N", "O", "F", "Ne",
    "Na", "Mg", "Al", "Si", "P", "S", "Cl", "Ar", "K", "Ca",
    "Sc", "Ti", "V", "Cr", "Mn", "Fe", "Co", "Ni", "Cu", "Zn",
    "Ga", "Ge", "As", "Se", "Br", "Kr", "Rb", "Sr", "Y", "Zr",
    "Nb", "Mo", "Tc", "Ru", "Rh", "Pd", "Ag", "Cd", "In", "Sn",
    "Sb", "Te", "I", "Xe", "Cs", "Ba", "La", "Ce", "Pr", "Nd",
    "Pm", "Sm", "Eu", "Gd", "Tb", "Dy", "Ho", "Er", "Tm", "Yb",
    "Lu", "Hf", "Ta", "W", "Re", "Os", "Ir", "Pt", "Au", "Hg",
    "Tl", "Pb", "Bi", "Po", "At", "Rn", "Fr", "Ra", "Ac", "Th",
    "Pa", "U", "Np", "Pu", "Am", "Cm", "Bk", "Cf", "Es", "Fm",
    "Md", "No", "Lr", "Rf", "Db", "Sg", "Bh", "Hs", "Mt", "Ds",
    "Rg", "Cn", "Nh", "Fl", "Mc", "Lv", "Ts", "Og"
]

COLOR_MAP = {'s': '#FF4136', 'p': '#FFDC00', 'd': '#0074D9', 'f': '#2ECC40'}

block_shapes = {
    's': ['1'],
    'p': ['z', 'x', 'y'],
    'd': ['2*z**2 - x**2 - y**2', 'x*z', 'y*z', 'x*y', 'x**2 - y**2'],
    'f': [
        'y * (3*x**2 - y**2)', 'x * (x**2 - 3*y**2)', 'z * (x**2 - y**2)',
        'z * (5*z**2 - 3)', 'x * (5*z**2 - 1)', 'y * (5*z**2 - 1)', 'x * y * z'
    ]
}


def generate_base_sphere(n_theta, n_phi):
    theta = np.linspace(0, np.pi, n_theta)
    phi   = np.linspace(0, 2*np.pi, n_phi)
    theta, phi = np.meshgrid(theta, phi)
    theta = theta.flatten()
    phi = phi.flatten()

    x = np.sin(theta) * np.cos(phi)
    y = np.sin(theta) * np.sin(phi)
    z = np.cos(theta)

    indices_i, indices_j, indices_k = [], [], []
    rows, cols = n_phi, n_theta

    for r in range(rows - 1):
        for c in range(cols - 1):
            p1 = r * cols + c
            p2 = r * cols + c + 1
            p3 = (r + 1) * cols + c
            p4 = (r + 1) * cols + c + 1
            indices_i.extend([p1, p3])
            indices_j.extend([p2, p2])
            indices_k.extend([p3, p4])

    return x, y, z, np.array(indices_i), np.array(indices_j), np.array(indices_k)


BASE_X, BASE_Y, BASE_Z, BASE_I, BASE_J, BASE_K = generate_base_sphere(N_THETA, N_PHI)

def get_deformed_mesh(expr, center, scale):
    SAFE = {'np': np, 'x': BASE_X, 'y': BASE_Y, 'z': BASE_Z}

    if expr == '1':
        f = np.ones_like(BASE_X)
    else:
        try:
            f = eval(expr, {'__builtins__': {}}, SAFE)
        except:
            f = BASE_X

    f_abs = np.abs(f)
    max_val = np.quantile(f_abs, 0.99) + 1e-9
    R = scale * (f_abs / max_val)

    X = R * BASE_X + center[0]
    Y = R * BASE_Y + center[1]
    Z = R * BASE_Z + center[2]

    return X, Y, Z, f_abs

def get_box_lines(center, size=0.95):
    cx, cy, cz = center
    r = size / 2



    x = [cx-r, cx+r, cx+r, cx-r, cx-r, cx-r, cx+r, cx+r, cx-r, cx-r, cx+r, cx+r]
    y = [cy-r, cy-r, cy+r, cy+r, cy-r, cy-r, cy-r, cy+r, cy+r, cy+r, cy+r, cy-r]
    z = [cz-r, cz-r, cz-r, cz-r, cz-r, cz+r, cz+r, cz+r, cz+r, cz-r, cz-r, cz+r]


    xs, ys, zs = [], [], []

    def add_line(p1, p2):
        xs.extend([p1[0], p2[0], None])
        ys.extend([p1[1], p2[1], None])
        zs.extend([p1[2], p2[2], None])


    def v(dx, dy, dz): return (cx+dx*r, cy+dy*r, cz+dz*r)


    for dx in [-1, 1]:
        for dy in [-1, 1]:
            add_line(v(dx, dy, -1), v(dx, dy, 1))
    for dx in [-1, 1]:
        for dz in [-1, 1]:
            add_line(v(dx, -1, dz), v(dx, 1, dz))
    for dy in [-1, 1]:
        for dz in [-1, 1]:
            add_line(v(-1, dy, dz), v(1, dy, dz))

    return xs, ys, zs



def get_atom_data_strict(z):

    if z in [1, 2]: period=1; block='s'
    elif z in [3,4]: period=2; block='s'
    elif 5<=z<=10: period=2; block='p'
    elif z in [11,12]: period=3; block='s'
    elif 13<=z<=18: period=3; block='p'
    elif z in [19,20]: period=4; block='s'
    elif 21<=z<=30: period=4; block='d'
    elif 31<=z<=36: period=4; block='p'
    elif z in [37,38]: period=5; block='s'
    elif 39<=z<=48: period=5; block='d'
    elif 49<=z<=54: period=5; block='p'
    elif z in [55,56]: period=6; block='s'
    elif 57<=z<=70: period=6; block='f'
    elif 71<=z<=80: period=6; block='d'
    elif 81<=z<=86: period=6; block='p'
    elif z in [87,88]: period=7; block='s'
    elif 89<=z<=102: period=7; block='f'
    elif 103<=z<=112: period=7; block='d'
    elif 113<=z<=118: period=7; block='p'
    else: return None

    z_pos = 8 - period

    start_z = 0
    if block == 's':
        start_z = [0, 1, 3, 11, 19, 37, 55, 87][period] if period < 8 else 0
    elif block == 'p':
        start_z = {2:5, 3:13, 4:31, 5:49, 6:81, 7:113}[period]
    elif block == 'd':
        start_z = {4:21, 5:39, 6:71, 7:103}[period]
    elif block == 'f':
        start_z = {6:57, 7:89}[period]

    rel_idx = z - start_z
    w_map = {'s':1, 'p':3, 'd':5, 'f':7}
    w = w_map[block]

    col_idx = rel_idx % w
    is_back = (rel_idx >= w)

    x_off = {'s':0, 'p':1, 'd':4, 'f':9}[block]
    x_pos = x_off + col_idx
    y_pos = 1 if is_back else 0

    return x_pos, y_pos, z_pos, block, rel_idx, period


merged_meshes = {
    's': {'x':[], 'y':[], 'z':[], 'i':[], 'j':[], 'k':[], 'v':[]},
    'p': {'x':[], 'y':[], 'z':[], 'i':[], 'j':[], 'k':[], 'v':[]},
    'd': {'x':[], 'y':[], 'z':[], 'i':[], 'j':[], 'k':[], 'v':[]},
    'f': {'x':[], 'y':[], 'z':[], 'i':[], 'j':[], 'k':[], 'v':[]}
}


box_lines = {'x': [], 'y': [], 'z': []}


label_data = {'x': [], 'y': [], 'z': [], 'text': []}

print("Computing High-Res Geometry & Grid Lines...")

for z in range(1, 119):
    res = get_atom_data_strict(z)
    if not res: continue
    vx, vy, vz, block, rel_idx, period = res
    sym = ALL_ELEMENTS[z]


    shapes = block_shapes[block]
    v_count = rel_idx + 1
    if sym in CONFIG_ANOMALIES:
        v_count = CONFIG_ANOMALIES[sym]
    shape_expr = shapes[(v_count - 1) % len(shapes)]

    scale = 0.20 + (period * 0.04)
    lx, ly, lz, lv = get_deformed_mesh(shape_expr, (vx, vy, vz), scale)

    target = merged_meshes[block]
    off = len(target['x'])
    target['x'].extend(lx); target['y'].extend(ly); target['z'].extend(lz)
    target['v'].extend(lv)
    target['i'].extend(BASE_I + off); target['j'].extend(BASE_J + off); target['k'].extend(BASE_K + off)


    bx, by, bz = get_box_lines((vx, vy, vz), size=0.95)
    box_lines['x'].extend(bx)
    box_lines['y'].extend(by)
    box_lines['z'].extend(bz)

    label_y = vy - 0.6 if vy == 0 else vy + 0.6
    label_data['x'].append(vx)
    label_data['y'].append(label_y)
    label_data['z'].append(vz)
    label_data['text'].append(sym)

fig = go.Figure()



for block in ['s', 'p', 'd', 'f']:
    d = merged_meshes[block]
    if not d['x']: continue
    fig.add_trace(go.Mesh3d(
        x=d['x'], y=d['y'], z=d['z'],
        i=d['i'], j=d['j'], k=d['k'],
        intensity=d['v'],
        colorscale=[[0, COLOR_MAP[block]], [1, 'white']],
        showscale=False,
        name=f"{block}-block",
        flatshading=False,
        lighting=dict(ambient=0.4, diffuse=0.7, roughness=0.1, specular=0.5),
        hoverinfo='skip'
    ))


fig.add_trace(go.Scatter3d(
    x=label_data['x'], y=label_data['y'], z=label_data['z'],
    mode='text',
    text=label_data['text'],
    textfont=dict(color='black', size=11),
    hoverinfo='text',
    name='Labels'
))



tick_vals = [1, 2, 3, 4, 5, 6, 7]
tick_text = ["n=7", "n=6", "n=5", "n=4", "n=3", "n=2", "n=1"]

fig.update_layout(
    title="Quantum Pyramid (High-Res & Wireframe Grid)",
    width=1200,
    height=900,


    paper_bgcolor='white',
    plot_bgcolor='white',

    scene=dict(
        bgcolor='white',

        xaxis=dict(
            visible=True,
            title="Orbital Type",
            showticklabels=False,
            showgrid=False,
            showbackground=False
        ),

        yaxis=dict(
            visible=True,
            title="Spin (Depth)",
            showticklabels=False,
            showgrid=False,
            showbackground=False
        ),

        zaxis=dict(
            visible=True,
            title="Energy Level (n)",
            tickmode='array',
            tickvals=tick_vals,
            ticktext=tick_text,
            showgrid=True,
            gridcolor='rgba(0,0,0,0.15)',
            showbackground=False
        ),

        camera=dict(
            eye=dict(x=2.2, y=-2.2, z=1.2),
            center=dict(x=0, y=0, z=-0.2)
        ),

        aspectmode='data'
    ),

    margin=dict(l=0, r=0, b=0, t=40)
)

filename = 'quantum_pyramid_highres.html'
fig.write_html(filename, config={'displayModeBar': False})
print(f"Done. High-Res model saved to {filename}")
fig.show()
```
