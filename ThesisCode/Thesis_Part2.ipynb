# Revealing Prime Structures via Primorial Filtering


**Part 1. Generate Prime Candidates**



*   Table only contains prime candidates




```python
import pandas as pd
from math import prod, sqrt
pd.set_option('display.max_columns', None)

def is_prime(n):
    """ Testing if a number is prime """
    if n < 2:
        return False
    if n in (2, 3):
        return True
    if n % 2 == 0 or n % 3 == 0:
        return False
    for i in range(5, int(sqrt(n)) + 1, 2):
        if n % i == 0:
            return False
    return True

def get_first_k_primes(k):
    """ Get first k prime numbers """
    primes = []
    num = 2
    while len(primes) < k:
        if is_prime(num):
            primes.append(num)
        num += 1
    return primes

def primorial(k):
    """ Calculate the kth Primorial Number """
    primes = get_first_k_primes(k)
    return prod(primes), primes


def find_candidate_primes(k):
    """ Finding prime numbers between R_k and R_(k+1) """
    R_k, primes_k = primorial(k)

    next_prime = primes_k[-1] + 1
    while not is_prime(next_prime):
        next_prime += 1
    R_k1 = R_k * next_prime

    candidate_offsets = [c for c in range(1, R_k, 2) if all(c % p != 0 for p in primes_k)]
    candidates = [R_k + c for c in candidate_offsets]

    new_primes = [n for n in candidates if is_prime(n)]

    return R_k, R_k1, candidate_offsets, new_primes

def structured_nk_c_table(k):
    """ Generate a structured nk + c format table with all candidate values """
    R_k, R_k1, candidate_offsets, _ = find_candidate_primes(k)


    num_rows = (R_k1 - R_k) // R_k +1


    table_data = [[R_k * i + c for c in candidate_offsets] for i in range(num_rows)]


    df_table = pd.DataFrame(table_data, columns=[f"{R_k}k + {c}" for c in candidate_offsets],
                            index=[f"k = {i}" for i in range(num_rows)])

    return df_table

def structured_nk_c_prime_table_numbers(k):
    """ Generate a table and correctly filter only prime candidates, displaying only numbers """
    R_k, R_k1, candidate_offsets, _ = find_candidate_primes(k)

    num_rows = (R_k1 - R_k) // R_k +1
    table_data = [[R_k * i + c if is_prime(R_k * i + c) else ""
                   for c in candidate_offsets] for i in range(num_rows)]

    df_prime_table = pd.DataFrame(table_data, columns=[f"{R_k}k + {c}" for c in candidate_offsets],
                                  index=[f"k = {i}" for i in range(num_rows)])

    return df_prime_table


k = 4
df_nk_c_full = structured_nk_c_table(k)
df_nk_c_prime_only = structured_nk_c_prime_table_numbers(k)


print("Complete nk + c Table (All Candidates):")
print(df_nk_c_full)
print("\nFiltered nk + c Prime-Only Table:")
print(df_nk_c_prime_only)

```

```python
import pandas as pd
from math import prod, sqrt

pd.set_option('display.max_columns', None)


def is_prime(n):
    if n < 2:
        return False
    if n in (2, 3):
        return True
    if n % 2 == 0 or n % 3 == 0:
        return False
    for i in range(5, int(sqrt(n)) + 1, 2):
        if n % i == 0:
            return False
    return True

def get_first_k_primes(k):
    primes = []
    num = 2
    while len(primes) < k:
        if is_prime(num):
            primes.append(num)
        num += 1
    return primes

def primorial(k):
    primes = get_first_k_primes(k)
    return prod(primes), primes

def find_candidate_primes(k):
    R_k, primes_k = primorial(k)

    next_prime = primes_k[-1] + 1
    while not is_prime(next_prime):
        next_prime += 1
    R_k1 = R_k * next_prime

    candidate_offsets = [c for c in range(1, R_k, 2) if all(c % p != 0 for p in primes_k)]
    candidates = [R_k + c for c in candidate_offsets]
    new_primes = [n for n in candidates if is_prime(n)]

    return R_k, R_k1, candidate_offsets, new_primes

def prime_factors(n):
    factors = []
    d = 2
    while d * d <= n:
        while n % d == 0:
            factors.append(d)
            n //= d
        d += 1
    if n > 1:
        factors.append(n)
    return ' × '.join(map(str, factors))


def structured_nk_c_prime_table_factors(k):
    R_k, R_k1, candidate_offsets, _ = find_candidate_primes(k)
    num_rows = (R_k1 - R_k) // R_k + 1

    table_data = []
    for i in range(num_rows):
        row = []
        for c in candidate_offsets:
            n = R_k * i + c
            if is_prime(n):
                row.append(str(n))
            else:
                row.append(prime_factors(n))
        table_data.append(row)

    df_table = pd.DataFrame(table_data, columns=[f"{R_k}k + {c}" for c in candidate_offsets],
                            index=[f"k = {i}" for i in range(num_rows)])
    return df_table



def print_box_title(title):
    line = "+" + "-" * (len(title) + 4) + "+"
    print(f"\n{line}\n|  {title}  |\n{line}\n")

if __name__ == "__main__":
    k = 4
    df_result = structured_nk_c_prime_table_factors(k)

    print_box_title(f"nk + c Table with Primes and Factorized Composites (k = {k})")
    print(df_result)
```

**Part 2. Evolution of Maximal Prime Gaps**

```python
import numpy as np
import plotly.graph_objects as go
import plotly.io as pio

# Set the base
PRIMORIAL_BASE = 210
ROWS_TO_DISPLAY = 11
SEARCH_LIMIT = PRIMORIAL_BASE * ROWS_TO_DISPLAY


def simple_sieve(limit):
    if limit < 2:
        return np.array([])
    is_prime = np.ones(limit + 1, dtype=bool)
    is_prime[0:2] = False
    for p in range(2, int(np.sqrt(limit)) + 1):
        if is_prime[p]:
            is_prime[p*p:limit+1:p] = False
    return np.nonzero(is_prime)[0]

def analyze_primes(primes, base):
    all_primes_k = primes // base
    all_primes_c = primes % base
    max_gaps_data = []
    current_max_gap = 0

    for i in range(len(primes) - 1):
        p1 = primes[i]
        p2 = primes[i + 1]
        gap = p2 - p1
        if gap > current_max_gap:
            current_max_gap = gap
            max_gaps_data.append({
                'gap': gap,
                'k': p1 // base,
                'c': p1 % base
            })

    return all_primes_k, all_primes_c, max_gaps_data

print(f"Calculating R={PRIMORIAL_BASE} ...")
primes = simple_sieve(SEARCH_LIMIT)
base_k, base_c, gap_records = analyze_primes(primes, PRIMORIAL_BASE)

gap_k = [r['k'] for r in gap_records]
gap_c = [r['c'] for r in gap_records]
gap_z = [r['gap'] for r in gap_records]


fig = go.Figure()

fig.add_trace(go.Scatter3d(
    x=base_k,
    y=base_c,
    z=np.zeros(len(base_k)),
    mode='markers',
    marker=dict(size=2, color='lightgray', opacity=0.3),
    name='Primes (Base)',
    hoverinfo='skip'
))


fig.add_trace(go.Scatter3d(
    x=gap_k,
    y=gap_c,
    z=gap_z,
    mode='lines',
    line=dict(color='rgba(80, 80, 80, 0.5)', width=3),
    name='Evolution Path'
))

stem_x, stem_y, stem_z = [], [], []
for k, c, z in zip(gap_k, gap_c, gap_z):
    stem_x.extend([k, k, None])
    stem_y.extend([c, c, None])
    stem_z.extend([0, z, None])

fig.add_trace(go.Scatter3d(
    x=stem_x,
    y=stem_y,
    z=stem_z,
    mode='lines',
    line=dict(color='rgba(255, 0, 0, 0.3)', width=1),
    showlegend=False
))

hover_texts = [
    f"Gap: {r['gap']}<br>Pos: {r['k']}*{PRIMORIAL_BASE} + {r['c']}"
    for r in gap_records
]

point_labels = [str(r['gap']) for r in gap_records]

fig.add_trace(go.Scatter3d(
    x=gap_k,
    y=gap_c,
    z=gap_z,
    mode='markers+text',
    marker=dict(
        size=5,
        color='royalblue',
        opacity=1.0
    ),
    text=point_labels,
    textposition='top center',
    textfont=dict(
        size=10,
        color='black'
    ),
    hovertext=hover_texts,
    hoverinfo='text',
    name='Max Gaps'
))


max_gap_val = max(gap_z) if gap_z else 10
x_len = ROWS_TO_DISPLAY
y_len = PRIMORIAL_BASE
z_len = max_gap_val * 1.1

axis_color = "#444444"
arrow_color = "#222222"
lw = 6

fig.add_trace(go.Scatter3d(
    x=[0, x_len], y=[0, 0], z=[0, 0],
    mode="lines",
    line=dict(color=axis_color, width=lw),
    showlegend=False
))
fig.add_trace(go.Scatter3d(
    x=[0, 0], y=[0, y_len], z=[0, 0],
    mode="lines",
    line=dict(color=axis_color, width=lw),
    showlegend=False
))
fig.add_trace(go.Scatter3d(
    x=[0, 0], y=[0, 0], z=[0, z_len],
    mode="lines",
    line=dict(color=axis_color, width=lw),
    showlegend=False
))


def add_arrow(tip, vec, length):
    ref_size = max(length * 0.1, 0.5)
    fig.add_trace(go.Cone(
        x=[tip[0]], y=[tip[1]], z=[tip[2]],
        u=[vec[0]], v=[vec[1]], w=[vec[2]],
        anchor="tip",
        sizemode="absolute",
        sizeref=ref_size,
        showscale=False,
        colorscale=[[0, arrow_color], [1, arrow_color]],
        hoverinfo="skip"
    ))

add_arrow((x_len, 0, 0), (1, 0, 0), length=x_len)
add_arrow((0, y_len, 0), (0, 1, 0), length=y_len)
add_arrow((0, 0, z_len), (0, 0, 1), length=z_len)


fig.update_layout(
    title=dict(text=f'Primorial Max Gaps (R={PRIMORIAL_BASE})', y=0.9),
    scene=dict(
        xaxis=dict(
            title='Row k',
            range=[0, x_len],
            backgroundcolor="rgb(245, 245, 245)"
        ),
        yaxis=dict(
            title='Offset c',
            range=[0, y_len],
            backgroundcolor="rgb(245, 245, 245)"
        ),
        zaxis=dict(
            title='Gap Size',
            range=[0, z_len],
            backgroundcolor="rgb(255, 255, 255)"
        ),
        aspectmode='cube'
    ),
    margin=dict(l=0, r=0, b=0, t=50),
    showlegend=False
)


output_filename = "primorial_max_gaps_labeled.html"
pio.write_html(fig, output_filename)
print(f"Save the file as {output_filename}")

fig.show()
```

**Part 3. k-tulpes Filter**

```python
from sympy import isprime

# construct the base
mod_base = 2310
offsets = [x for x in range(mod_base) if all(x % p != 0 for p in [2, 3, 5, 7])]

candidates = []
for k in range(2310 // mod_base + 1):
    for off in offsets:
        n = k * mod_base + off
        if n <= 2310:
            candidates.append(n)

primes = sorted([n for n in candidates if isprime(n)])


tuple_patterns = [
    [0, 2],
    [0, 2, 6],
    [0, 4, 6],
    [0, 2, 6, 8],
    [0, 2, 6, 8, 12],
    [0, 4, 6, 10, 12],
    [0, 4, 6, 10, 12, 16],
    [0, 2, 6, 8, 12, 18, 20],
    [0, 2, 8, 12, 14, 18, 20],
    [0, 2, 6, 12, 14, 20, 24, 26],
    [0, 6, 8, 14, 18, 20, 24, 26],
    [0, 2, 6, 8, 12, 18, 20, 26, 30],
    [0, 4, 6, 10, 16, 18, 24, 28, 30],
    [0, 2, 6, 12, 14, 20, 24, 26, 30],
    [0, 4, 10, 12, 18, 22, 24, 29, 30],

]


def match_pattern(prime_list, pattern):
    matches = []
    prime_set = set(prime_list)
    for p in prime_list:
        if all((p + d) in prime_set for d in pattern):
            matches.append([p + d for d in pattern])
    return matches

for pattern in sorted(tuple_patterns, key=len, reverse=True):  # 优先查长的
    result = match_pattern(primes, pattern)
    if result:
        print("=" * 50)
        print(f"Found {len(result)} matches for pattern {pattern}:")
        for m in result:
            print(m)

```

**Showing The Partial Order Set of K-tuples**

```python
import math
import random
from itertools import combinations
import networkx as nx
import plotly.graph_objects as go


tuple_patterns = [
    [0, 2],
    [0, 2, 6], [0, 4, 6],
    [0, 2, 6, 8], [0, 4, 6, 10],
    [0, 2, 6, 8, 12], [0, 4, 6, 10, 12],
    [0, 4, 6, 10, 12, 16],
    [0, 2, 6, 8, 12, 18, 20], [0, 2, 8, 12, 14, 18, 20],
    [0, 2, 6, 12, 14, 20, 24, 26], [0, 6, 8, 14, 18, 20, 24, 26],
    [0, 2, 6, 8, 12, 18, 20, 26, 30], [0, 4, 6, 10, 16, 18, 24, 28, 30],
    [0, 2, 6, 12, 14, 20, 24, 26, 30], [0, 4, 10, 12, 18, 22, 24, 28, 30],
]

def name_for(p): return "[" + ", ".join(str(x) for x in p) + "]"

patterns_set = {name_for(p): set(p) for p in tuple_patterns}
rank = {n: len(p) for n, p in patterns_set.items()}
span_d = {n: (max(p) - min(p)) if p else 0 for n, p in patterns_set.items()}


base_palette = {
    "[0, 2]":"#1F77B4", "[0, 2, 6]":"#9467BD", "[0, 4, 6]":"#8C564B",
    "[0, 2, 6, 8]":"#D62728", "[0, 4, 6, 10]":"#000000",
    "[0, 2, 6, 8, 12]":"#BCBD22", "[0, 4, 6, 10, 12]":"#8C564B",
    "[0, 4, 6, 10, 12, 16]":"#FFBB78", "[0, 2, 6, 8, 12, 18, 20]":"#1CA9C9",
    "[0, 2, 8, 12, 14, 18, 20]":"#E377C2", "[0, 2, 6, 12, 14, 20, 24, 26]":"#AEC7E8",
    "[0, 6, 8, 14, 18, 20, 24, 26]":"#C23B22", "[0, 2, 6, 8, 12, 18, 20, 26, 30]":"#2E8B57",
    "[0, 4, 6, 10, 16, 18, 24, 28, 30]":"#7F7F7F", "[0, 2, 6, 12, 14, 20, 24, 26, 30]":"#9EDAE5",
    "[0, 4, 10, 12, 18, 22, 24, 28, 30]":"#9467BD",
}
symbol_char_map = {2: "★", 3: "♦", 4: "♦", 5: "♣", 6: "♣", 7: "♥", 8: "♥", 9: "♠"}


nodes = list(patterns_set.keys())
G_strict = nx.DiGraph()
G_translation = nx.DiGraph()

for a, b in combinations(nodes, 2):
    ka, kb = rank[a], rank[b]
    u, v = (a, b) if ka < kb else (b, a)
    U, V = patterns_set[u], patterns_set[v]

    if U.issubset(V):
        G_strict.add_edge(u, v)


    for t in range(-30, 31):
        if {x + t for x in U}.issubset(V):
            G_translation.add_edge(u, v)
            break


H_strict = nx.algorithms.dag.transitive_reduction(G_strict)
H_trans  = nx.algorithms.dag.transitive_reduction(G_translation)


DZ = 1.8
OVERRIDES = {
    "[0, 2]": (0.0, 0.0), "[0, 2, 6]": (2.0, 0.2), "[0, 4, 6]": (-2.0, 0.4),
    "[0, 2, 6, 8]": (3.5, 1.2), "[0, 4, 6, 10]": (-3.5, 1.2),
    "[0, 2, 6, 8, 12]": (4.0, 1.4), "[0, 4, 6, 10, 12]": (-4.0, 1.0),
    "[0, 4, 6, 10, 12, 16]": (-4.0, 1.8), "[0, 2, 6, 8, 12, 18, 20]": (4.0, 2.5),
    "[0, 2, 8, 12, 14, 18, 20]": (0.2, 0.2), "[0, 2, 6, 12, 14, 20, 24, 26]": (2.0, 2.6),
    "[0, 6, 8, 14, 18, 20, 24, 26]": (-1.2, 2.5), "[0, 2, 6, 8, 12, 18, 20, 26, 30]": (5.0, 1.4),
    "[0, 4, 6, 10, 16, 18, 24, 28, 30]": (-3.4, 1.0), "[0, 2, 6, 12, 14, 20, 24, 26, 30]": (3.1, 3.0),
    "[0, 4, 10, 12, 18, 22, 24, 28, 30]": (-1.5, 3.3),
}
sorted_k = sorted(list(set(rank.values())))
k_to_idx = {k: i for i, k in enumerate(sorted_k)}
pos3d = {n: (OVERRIDES[n][0], OVERRIDES[n][1], k_to_idx[rank[n]] * DZ) for n in nodes}


fig = go.Figure()

for i, k in enumerate(sorted_k):
    z, s = i * DZ, 2.5 + 0.4 * i
    fig.add_trace(go.Surface(x=[[-s, s], [-s, s]], y=[[-s, -s], [s, s]], z=[[z, z], [z, z]],
                             showscale=False, opacity=0.04, colorscale=[[0, '#CCC'], [1, '#CCC']], hoverinfo="skip"))

sx, sy, sz = [], [], []
cone_s_x, cone_s_y, cone_s_z = [], [], []
cone_s_u, cone_s_v, cone_s_w = [], [], []

for u, v in H_strict.edges():
    p0, p1 = pos3d[u], pos3d[v]

    sx += [p0[0], p1[0], None]; sy += [p0[1], p1[1], None]; sz += [p0[2], p1[2], None]

    vx, vy, vz = p1[0]-p0[0], p1[1]-p0[1], p1[2]-p0[2]

    length = math.sqrt(vx**2 + vy**2 + vz**2)
    if length > 0:
        vx, vy, vz = vx/length, vy/length, vz/length

    cone_s_x.append(p0[0] + (p1[0]-p0[0]) * 0.55)
    cone_s_y.append(p0[1] + (p1[1]-p0[1]) * 0.55)
    cone_s_z.append(p0[2] + (p1[2]-p0[2]) * 0.55)


    cone_s_u.append(vx)
    cone_s_v.append(vy)
    cone_s_w.append(vz)


fig.add_trace(go.Scatter3d(x=sx, y=sy, z=sz, mode="lines",
                           line=dict(color="#000000", width=4.5),
                           name="Strict Inclusion (Black Solid)"))

fig.add_trace(go.Cone(x=cone_s_x, y=cone_s_y, z=cone_s_z, u=cone_s_u, v=cone_s_v, w=cone_s_w,
                      showscale=False, sizemode="absolute", sizeref=0.3, anchor="center",
                      colorscale=[[0, 'black'], [1, 'black']], hoverinfo="skip"))

tx, ty, tz = [], [], []
cone_t_x, cone_t_y, cone_t_z = [], [], []
cone_t_u, cone_t_v, cone_t_w = [], [], []
strict_edges = set(H_strict.edges())

for u, v in H_trans.edges():
    if (u, v) not in strict_edges:
        p0, p1 = pos3d[u], pos3d[v]
        tx += [p0[0], p1[0], None]; ty += [p0[1], p1[1], None]; tz += [p0[2], p1[2], None]

        vx, vy, vz = p1[0]-p0[0], p1[1]-p0[1], p1[2]-p0[2]

        length = math.sqrt(vx**2 + vy**2 + vz**2)
        if length > 0:
            vx, vy, vz = vx/length, vy/length, vz/length

        cone_t_x.append(p0[0] + (p1[0]-p0[0]) * 0.55)
        cone_t_y.append(p0[1] + (p1[1]-p0[1]) * 0.55)
        cone_t_z.append(p0[2] + (p1[2]-p0[2]) * 0.55)

        cone_t_u.append(vx)
        cone_t_v.append(vy)
        cone_t_w.append(vz)


fig.add_trace(go.Scatter3d(x=tx, y=ty, z=tz, mode="lines",
                           line=dict(color="#888888", width=2.5, dash="dash"),
                           name="Translation Relation (Grey Dash)"))

fig.add_trace(go.Cone(x=cone_t_x, y=cone_t_y, z=cone_t_z, u=cone_t_u, v=cone_t_v, w=cone_t_w,
                      showscale=False, sizemode="absolute", sizeref=0.3, anchor="center",
                      colorscale=[[0, '#888888'], [1, '#888888']], hoverinfo="skip"))

node_x, node_y, node_z = [pos3d[n][0] for n in nodes], [pos3d[n][1] for n in nodes], [pos3d[n][2] for n in nodes]
fig.add_trace(go.Scatter3d(x=node_x, y=node_y, z=node_z, mode="markers+text",
    text=[symbol_char_map.get(rank[n], "●") for n in nodes],
    textposition="middle center",
    textfont=dict(size=18, color=[base_palette.get(n, "#333") for n in nodes]),
    marker=dict(size=12, color=[base_palette.get(n, "#333") for n in nodes], opacity=0.22),
    hovertext=[f"{n}<br>k={rank[n]}, d={span_d[n]}" for n in nodes], hoverinfo="text", showlegend=False))

fig.add_trace(go.Scatter3d(x=node_x, y=node_y, z=[z + 0.42 for z in node_z], mode="text",
    text=[f"{n}<br>k={rank[n]}, d={span_d[n]}" for n in nodes],
    textfont=dict(size=10, color="#333"), textposition="top center", showlegend=False))


fig.update_layout(
    scene=dict(xaxis=dict(visible=False), yaxis=dict(visible=False),
               zaxis=dict(title="Order (k)", tickmode="array",
                         tickvals=[i * DZ for i in range(len(sorted_k))],
                         ticktext=[str(k) for k in sorted_k],
                         gridcolor="#EEE"),
               camera=dict(eye=dict(x=1.6, y=1.6, z=1.0)),
               bgcolor="white"),
    paper_bgcolor="white",
    margin=dict(l=0, r=0, t=30, b=0),
    showlegend=True
)
file_name = "POSET_Diagram_Consistent_Arrow.html"
fig.write_html(file_name, include_plotlyjs='cdn')
fig.show()
```

**Part 4. 3D visualization of k-tuples distribution**

```python
import plotly.graph_objects as go
import plotly.io as pio
from sympy import isprime


base_palette = {
    "[0,2]": "#1F77B4",
    "[0,2,6]": "#9467BD",
    "[0,4,6]": "#8C564B",
    "[0,2,6,8]": "#D62728",
    "[0,4,6,10]": "#000000",
    "[0,2,6,8,12]": "#D62728",
    "[0,4,6,10,12]": "#8C564B",
    "[0,4,6,10,12,16]": "#FFBB78",
    "[0,2,6,8,12,18,20]": "#D62728",
    "[0,2,8,12,14,18,20]": "#ADD8E6",
    "[0,2,6,12,14,20,24,26]": "#B0C4DE",
    "[0,6,8,14,18,20,24,26]": "#FFD700",
    "[0,2,6,8,12,18,20,26,30]": "#D62728",
    "[0,4,6,10,16,18,24,28,30]": "#8C564B",
    "[0,2,6,12,14,20,24,26,30]": "#9467BD",
    "[0,4,10,12,18,22,24,28,30]": "#2E8B57",
}

def get_symbol_char(k):
    if k == 2: return "●"
    if k in [3, 4]: return "♦"
    if k in [5, 6]: return "♣"
    if k in [7, 8]: return "♥"
    if k >= 9: return "♠"
    return "■"

tuple_patterns = [
    [0, 2],
    [0, 2, 6], [0, 4, 6],
    [0, 2, 6, 8],
    [0, 2, 6, 8, 12], [0, 4, 6, 10, 12],
    [0, 4, 6, 10, 12, 16],
    [0, 2, 6, 8, 12, 18, 20], [0, 2, 8, 12, 14, 18, 20],
    [0, 2, 6, 12, 14, 20, 24, 26], [0, 6, 8, 14, 18, 20, 24, 26],
    [0, 2, 6, 8, 12, 18, 20, 26, 30], [0, 4, 6, 10, 16, 18, 24, 28, 30],
    [0, 2, 6, 12, 14, 20, 24, 26, 30], [0, 4, 10, 12, 18, 22, 24, 28, 30],
]

mod_base = 210
max_n = 4000
candidates = [n for n in range(max_n) if isprime(n)]
candidates_set = set(candidates)

def match_pattern(pattern):
    matched = []
    for p in candidates:
        if all((p + d in candidates_set) for d in pattern):
            matched.append(p)
    return matched

fig = go.Figure()

for pattern in tuple_patterns:
    p_str = "[" + ",".join(str(x) for x in pattern) + "]"
    starts = match_pattern(pattern)
    count = len(starts)

    if count == 0: continue

    k = len(pattern)
    color = base_palette.get(p_str, "#333333")
    symbol = get_symbol_char(k)

    xs = [p % mod_base for p in starts]
    ys = [p // mod_base for p in starts]
    zs = [k for _ in starts]

    display_name = f"{symbol} {p_str} ({count})"

    fig.add_trace(go.Scatter3d(
        x=xs, y=ys, z=zs,
        mode="markers+text",
        name=display_name,
        text=[symbol] * count,
        textposition="middle center",
        textfont=dict(size=22, color=color),
        marker=dict(
            size=6,
            color=color,
            opacity=0.01
        ),
        hoverinfo="text",
        hovertext=[f"Pattern: {p_str}<br>Prime: {p}<br>Count: {count}" for p in starts]
    ))

x_len, y_len, z_len = 220, 20, 10
axis_color, arrow_color, lw = "#444444", "#222222", 8

fig.add_trace(go.Scatter3d(x=[0, x_len], y=[0, 0], z=[0, 0], mode="lines", line=dict(color=axis_color, width=lw), showlegend=False))
fig.add_trace(go.Scatter3d(x=[0, 0], y=[0, y_len], z=[0, 0], mode="lines", line=dict(color=axis_color, width=lw), showlegend=False))
fig.add_trace(go.Scatter3d(x=[0, 0], y=[0, 0], z=[0, z_len], mode="lines", line=dict(color=axis_color, width=lw), showlegend=False))

def add_arrow(tip, vec, length):
    fig.add_trace(go.Cone(x=[tip[0]], y=[tip[1]], z=[tip[2]], u=[vec[0]], v=[vec[1]], w=[vec[2]],
                         anchor="tip", sizemode="absolute", sizeref=max(length*0.08, 0.8),
                         showscale=False, colorscale=[[0, arrow_color], [1, arrow_color]], hoverinfo="skip"))

add_arrow((x_len, 0, 0), (1, 0, 0), x_len)
add_arrow((0, y_len, 0), (0, 1, 0), y_len)
add_arrow((0, 0, z_len), (0, 0, 1), z_len)

fig.update_layout(
    scene=dict(
        xaxis=dict(
            title="Offset c (mod 210)",
            range=[0, x_len],
            showbackground=False,
            showgrid=False,
            showline=True,
            zeroline=False
        ),
        yaxis=dict(
            title="Layer n (layer in nk + c)",
            range=[0, y_len],
            showbackground=False,
            showgrid=False,
            showline=True,
            zeroline=False
        ),
        zaxis=dict(
            title="Tuple Length k",
            range=[0, z_len],
            showbackground=False,
            showgrid=False,
            showline=True,
            zeroline=False
        ),
        bgcolor="white",
        aspectmode="cube"
    ),
    paper_bgcolor="white",
    legend=dict(
        title="<b>Symbol | Pattern | Occurrence</b>",
        itemsizing="constant",
        font=dict(size=12, family="Courier New, monospace"),
        bgcolor="rgba(255,255,255,0.85)",
        bordercolor="Gray",
        borderwidth=1,
        traceorder="normal"
    ),
    margin=dict(l=0, r=0, t=50, b=0),
    title="3D Mapping of Prime k-tuples "
)

pio.write_html(fig, "TUPLES_WITH_LEGEND_SYMBOLS.html")
fig.show()
```
